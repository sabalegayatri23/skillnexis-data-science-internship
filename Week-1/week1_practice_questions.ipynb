```python
# ============================================================
# SKILLNEXIS DATA SCIENCE INTERNSHIP
# WEEK 1 - EXCEL & DATA CLEANING
# PRACTICE QUESTIONS
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

print("Libraries imported successfully!")
```

```python
# ============================================================
# LOAD DATASET
# ============================================================

# Find Excel files inside the current Week-1 folder
excel_files = list(Path.cwd().rglob("*.xlsx"))

print("Excel files found:")

for file in excel_files:
    print(file)

# Find Global Superstore dataset
file_path = None

for file in excel_files:
    if "superstore" in file.name.lower():
        file_path = file
        break

if file_path is None:
    raise FileNotFoundError(
        "Global Superstore Excel dataset was not found. "
        "Make sure the Excel file is inside the Week-1 folder."
    )

print("\nDataset selected:")
print(file_path)

# Load Orders sheet
df = pd.read_excel(file_path, sheet_name="Orders")

print("\n" + "=" * 60)
print("DATASET LOADED SUCCESSFULLY")
print("=" * 60)

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())
```

```python
# ============================================================
# BASIC DATASET INFORMATION
# ============================================================

print("Dataset Shape:")
print(df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())
```

```python
# ============================================================
# PRACTICE QUESTION 1
# REMOVE DUPLICATES
# ============================================================

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

if duplicate_count > 0:
    print("\nRemoving duplicate rows...")
    df = df.drop_duplicates().copy()
    print("Duplicates removed successfully.")
else:
    print("No duplicate rows found.")
    
print("\nRows after duplicate removal:", len(df))
```

```python
# ============================================================
# CHECK MISSING VALUES
# ============================================================

missing_values = df.isnull().sum()

print("Missing values in each column:")
display(missing_values[missing_values > 0])

total_missing = df.isnull().sum().sum()

print("\nTotal missing values:", total_missing)
```

```python
# ============================================================
# HANDLE MISSING VALUES
# ============================================================

# Numeric columns
numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].median())

# Categorical/text columns
categorical_columns = df.select_dtypes(include=["object"]).columns

for column in categorical_columns:
    if df[column].isnull().sum() > 0:
        mode_value = df[column].mode()

        if not mode_value.empty:
            df[column] = df[column].fillna(mode_value[0])
        else:
            df[column] = df[column].fillna("Unknown")

print("Missing values handled successfully!")

print("\nRemaining missing values:")
display(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
# ============================================================
# SAVE CLEANED DATA
# ============================================================

cleaned_file = Path.cwd() / "cleaned_sales_data.csv"

df.to_csv(cleaned_file, index=False)

print("Cleaned dataset saved as:")
print(cleaned_file)
```

```python
# ============================================================
# PRACTICE QUESTION 2
# REGION-WISE SALES PIVOT TABLE
# ============================================================

region_pivot = pd.pivot_table(
    df,
    index="Region",
    values="Sales",
    aggfunc="sum"
)

region_pivot = region_pivot.sort_values(
    by="Sales",
    ascending=False
)

print("REGION-WISE SALES")
print("=" * 50)

display(region_pivot)
```

```python
# ============================================================
# SAVE REGION-WISE SALES
# ============================================================

region_pivot.to_csv(
    "region_wise_sales.csv"
)

print("Region-wise sales saved as:")
print("region_wise_sales.csv")
```

```python
# ============================================================
# PRACTICE QUESTION 3
# MEAN, MEDIAN AND MODE
# ============================================================

sales = df["Sales"]

mean_sales = sales.mean()
median_sales = sales.median()
mode_sales = sales.mode()

print("=" * 60)
print("DESCRIPTIVE STATISTICS - SALES")
print("=" * 60)

print(f"Mean Sales   : ${mean_sales:,.2f}")
print(f"Median Sales : ${median_sales:,.2f}")

if not mode_sales.empty:
    print(f"Mode Sales   : ${mode_sales.iloc[0]:,.2f}")
else:
    print("Mode Sales   : No mode found")
```

```python
# ============================================================
# VARIANCE
# ============================================================

variance_sales = sales.var()

print(f"Variance of Sales : {variance_sales:,.2f}")
```

```python
# ============================================================
# COMPLETE DESCRIPTIVE STATISTICS TABLE
# ============================================================

statistics = pd.DataFrame({
    "Statistic": [
        "Mean",
        "Median",
        "Mode",
        "Variance"
    ],
    "Value": [
        mean_sales,
        median_sales,
        mode_sales.iloc[0] if not mode_sales.empty else np.nan,
        variance_sales
    ]
})

display(statistics)
```

```python
# ============================================================
# PRACTICE QUESTION 4
# TOP 10 SALES
# ============================================================

top_10_sales = df.nlargest(10, "Sales").copy()

print("=" * 60)
print("TOP 10 SALES")
print("=" * 60)

display(top_10_sales[[
    "Order ID",
    "Order Date",
    "Customer Name",
    "Category",
    "Sales"
]])
```

```python
# ============================================================
# HIGHLIGHT TOP 10 SALES
# ============================================================

top_10_display = top_10_sales[[
    "Order ID",
    "Order Date",
    "Customer Name",
    "Category",
    "Sales"
]].style.highlight_max(
    subset=["Sales"],
    color="yellow"
)

display(top_10_display)
```

```python
# ============================================================
# SAVE TOP 10 SALES
# ============================================================

top_10_sales.to_csv(
    "top_10_sales.csv",
    index=False
)

print("Top 10 sales saved as:")
print("top_10_sales.csv")
```

```python
# ============================================================
# PRACTICE QUESTION 5
# CATEGORY-WISE REVENUE
# ============================================================

category_revenue = df.groupby(
    "Category"
)["Sales"].sum().sort_values(
    ascending=False
)

print("=" * 60)
print("CATEGORY-WISE REVENUE")
print("=" * 60)

display(category_revenue)
```

```python
# ============================================================
# CATEGORY-WISE REVENUE BAR CHART
# ============================================================

plt.figure(figsize=(10, 6))

category_revenue.plot(
    kind="bar"
)

plt.title("Category-wise Revenue")
plt.xlabel("Category")
plt.ylabel("Total Revenue ($)")

plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

```python
# ============================================================
# SAVE CATEGORY-WISE REVENUE
# ============================================================

category_revenue.to_csv(
    "category_wise_revenue.csv"
)

print("Category-wise revenue saved successfully!")
```
